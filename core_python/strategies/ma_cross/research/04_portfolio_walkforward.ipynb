# 04 Walk-Forward - MA Cross

Run simple IS/OOS walk-forward checks for MA Cross and inspect stability across windows.


```python
# Cell 1 - Safe import path bootstrap

import sys
from pathlib import Path


def find_root(start: Path) -> Path:
    for p in [start, *start.parents]:
        if (p / 'pyproject.toml').exists() and (p / 'core_python' / 'shared').exists():
            return p
    raise RuntimeError('Cannot find SEN05 repo root')


ROOT = find_root(Path.cwd())
CORE = ROOT / 'core_python'
for p in (str(ROOT), str(CORE)):
    if p not in sys.path:
        sys.path.insert(0, p)

print('ROOT =', ROOT)
```

```python
# Cell 2 - Imports and notebook setup

from IPython.display import display

from core_python.strategies.ma_cross.research_utils import (
    configure_notebook,
    export_research_bundle,
    plot_walkforward_dashboard,
    show_run_config,
    show_strategy_summary,
    simple_walkforward,
    summarize_walkforward,
)

configure_notebook()
show_strategy_summary()
```

```python
# Cell 3 - Walk-forward configuration

RUN_CONFIG = {
    'symbol': 'US30',
    'tf': 'M30',
    'account_mode': 'standard',
    'initial_balance': 100_000.0,
    'max_bars': 80_000,
    'indicator_overrides': {},
    'strategy_overrides': {},
    'costs': {},
    'broker_profile': None,
    'windows': [
        ('2021-01-01', '2022-12-31', '2023-01-01', '2023-06-30'),
        ('2021-07-01', '2023-06-30', '2023-07-01', '2023-12-31'),
        ('2022-01-01', '2023-12-31', '2024-01-01', '2024-06-30'),
    ],
    'export_report': False,
}
show_run_config('MA Cross Walk-Forward Configuration', RUN_CONFIG)
```

```python
# Cell 4 - Run walk-forward

wf = simple_walkforward(**{k: v for k, v in RUN_CONFIG.items() if k != 'export_report'})
summarize_walkforward(wf)
plot_walkforward_dashboard(wf)
```

```python
# Cell 5 - Optional export

if RUN_CONFIG.get('export_report'):
    export_path = export_research_bundle({'walkforward': wf}, name='ma_cross_walkforward')
    print('Exported:', export_path)
else:
    print("Export is disabled. Set RUN_CONFIG['export_report'] = True to save CSV files.")
```
